```python
import yfinance as yf
import openpyxl
import pandas as pd
from datetime import date
datos_yahoo = pd.read_excel("../config/SP500_yfinance.xlsx")

sectores = datos_yahoo.groupby("Sector")["Ticket"].apply(list).to_dict()
acciones = datos_yahoo["Ticket"].to_list()
precios = yf.download(acciones, start="2026-01-01", end=str(date.today()), progress=False)["Close"]
```

```python
def datos_sector(ticket_sector,df_precios):
    tickets = [t for t in ticket_sector if t in df_precios.columns]
    return df_precios[tickets]
```

```python
datos_por_sector ={}
rendimientos_por_sector = {}
covarianzas_por_sector = {}

for sector, tickets in sectores.items():
    precio = datos_sector(tickets,precios)
    datos_por_sector[sector]=precio

    if not precio.empty:
        rendimientos = precio.pct_change().dropna()
        rendimientos_por_sector[sector]=rendimientos

        covarianzas_por_sector[sector]=rendimientos.cov()*252
```
